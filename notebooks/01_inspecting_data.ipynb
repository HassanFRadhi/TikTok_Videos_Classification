### **Part 2c. Understanding the data - Investigating the variables**

In this stage, we'll will begin to investigate the variables more closely to better understand them.

We know from the project proposal that the ultimate objective is to use machine learning to classify videos as either claims or opinions. A good first step towards understanding the data might therefore be examining the `claim_status` variable. we'll begin by determining how many videos there are for each different claim status.

```python
# The different values for claim status and how many of each are in the data
data['claim_status'].value_counts(normalize=True)
```

> We notice that the percentages are approximately equal.

Next, we'll examine the engagement trends associated with each different claim status.

This can be dine by using boolean masking to filter the data according to claim status, then calculating the mean and median view counts for each claim status.

```python
# The average view count of videos with "claim" status
print(data[data['claim_status'] == 'claim']['video_view_count'].mean())
data[data['claim_status'] == 'claim']['video_share_count'].median()
```

```python
# The average view count of videos with "opinion" status
print(data[data['claim_status'] == 'opinion']['video_view_count'].mean())
data[data['claim_status'] == 'opinion']['video_view_count'].median()
```

We notice that the median and mean for each indivisual value are approximately the same. However, there is a huge differance in the average views when the status is different, which indicates an obvious relation. <br>

Now, we'll examine trends associated with the ban status of the author using `groupby()` to calculate how many videos there are for each combination of categories of claim status and author ban status.

```python
# Counts for each group combination of claim status and author ban status
data.groupby(['claim_status', 'author_ban_status'], as_index=False).size()
```

We notice that there is a direct relation between video being `claim` and author being `banned`, since banned videos at the `claim` status are much higher than in `opinion`. The reason of this might be due to the fact that banned authors content depends highly on claims for aquiring more traffic.

We,ll ontinue investigating engagement levels focusing on `author_ban_status`.

Let's calculate the median video shares count of each author ban status.

```python
data.groupby('author_ban_status')['video_share_count'].median()
```

It's obious that banned users have much more shares on their vidoes, which indicated an important correlation. Also, `claim` videos appears to have much more shares.

Using `groupby()` to group the data by `author_ban_status`, then using `agg()`, we can get the count, mean, and median of each of the following columns:
* `video_view_count`
* `video_like_count`
* `video_share_count`


```python
data.groupby('author_ban_status')[['video_view_count','video_like_count','video_share_count']].agg(['count','mean','median'])
```

We can see that banned authore have much more views, likes and shares on ther videos compared to active uers.

Now, let's create three new columns to better understand engagement rates:
* `likes_per_view`: the number of likes divided by the number of views for each video
* `comments_per_view`: the number of comments divided by the number of views for each video
* `shares_per_view`: the number of shares divided by the number of views for each video

```python
# Creating a likes_per_view column
data['likes_per_view'] = data['video_like_count']/data['video_view_count']

# Creating a comments_per_view column
data['comments_per_view'] = data['video_comment_count']/data['video_view_count']
# Creating a shares_per_view column
data['shares_per_view'] = data['video_share_count']/data['video_view_count']
```

Using`groupby()` to compile the information in each of the three newly created columns for each combination of categories of claim status and author ban status, then using `agg()` to calculate the count, the mean, and the median of each group. we get:

```python
data.groupby(['claim_status','author_ban_status'])[['likes_per_view','comments_per_view','shares_per_view']].agg(['count','mean','median'])
```

`opinion` videos seem to have more active authors with less views, shares, comments and likes per view compared to `claim` videos but those percentages here are higher for active authors. <br>
On the other hand, `claim` videos seem to have more banned authors with higher views, shares, comments and likes per view compared to `opinion` videos and those percentages here are higher for banned authors.
